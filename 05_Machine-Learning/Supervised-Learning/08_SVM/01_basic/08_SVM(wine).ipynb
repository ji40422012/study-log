```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.family'] = 'Malgun Gothic'
#1. 표준화
#2. 하이퍼파라미터 튜닝: C:오차허용(작으면 여백^, 크면 과적합), gamma: 크면 구불구불
#3. 평가
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
wine = load_wine()
x, y = wine.data, wine.target
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state=42, stratify=y)
print(f'훈련: {x_train.shape}, 테스트:{x_test.shape}, 품종:{list(wine.target_names)}')
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'svc__C':[0.1,1,10,100],  # Pipeline의 svc 단계에 있는 C 파라미터
    'svc__gamma':['scale', 0.01, 0.1, 1],
    'svc__kernel':['linear', 'rbf']}
pipe = Pipeline([('scaler', StandardScaler()), ('svc', SVC())])
grid = GridSearchCV(pipe, param_grid, cv=5, scoring='accuracy', n_jobs=1)
grid.fit(x_train, y_train)

print('가장좋은설정:', grid.best_params_)
print('교차검증 정확도:', round(grid.best_score_,3))
print('테스트 정확도:', round(grid.score(x_test, y_test), 3))
```
